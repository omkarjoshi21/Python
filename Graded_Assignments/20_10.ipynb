```python
# Q1. Function Scope and Built-in Functions
def analyze_numbers(num):
    print("maximum=",max(num))
    print("minimum=",min(num))
    print("sum=",sum(num))
    print("average=",sum(num)/len(num))
analyze_numbers([10, 20, 30, 40])
```

```python
# Q2. Lambda Function
square=lambda n:n*n
n=float(input("enter a number :"))
print(square(n))
```

```python
# Q3. Math and Random Module
import math
import random
num = random.randint(1, 100)
print("random number:",num)
print("square root:",math.sqrt(num))
```

```python
# Q4. Employee Salary Calculator using Function Arguments
def calculate_salary(basic_salary,bonus=2000):
    return basic_salary + bonus

#default argument
print("Employee1 Salary=",calculate_salary(30000))
#positional arguments
print("Employee2 Salary=",calculate_salary(30000,5000))
#keyword arguments
print("Employee3 Salary=",calculate_salary(basic_salary=25000,bonus=2000))
```

```python
#Q6. Decorator Implementation
def my_decorator(func):
    def hi(name):
        print("Welcome")
        func(name)
    return hi

@my_decorator
def hel(name):
    print("Hello " + name)

hel("Omkar")
```

```python
# Q7. Generator and Datetime Module
from datetime import date, timedelta
def next():
    today = date.today()
    for i in range(7):
        yield today + timedelta(days=i)
for d in next():
    print(d)
```
